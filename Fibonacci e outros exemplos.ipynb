```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Используем генератор
for num in fibonacci(5):
    print(num)
```

Neste exemplo, o gerador Fibonacci calcula números sob demanda em vez de armazenar todos os valores na memória. Isto torna os geradores especialmente adequados para trabalhar com fluxos de dados ou sequências infinitas.

# **Outro modo de fazer o gerador**

```python

def factorial(n):
    if n == 0:
        return 1  # Базовый случай
    else:
        return n * factorial(n - 1)  # Рекурсивный вызов

print(factorial(5))  # Вывод: 120
```

É importante garantir que o caso base esteja definido corretamente, caso contrário a função poderá causar um estouro de pilha. A recursão é útil para problemas que podem ser divididos em subproblemas, como travessia de árvore ou resolução de problemas de sequência.

**Um construtor em Python** é um método especial de uma classe que é chamado automaticamente quando um novo objeto dessa classe é criado. Em Python, um construtor é indicado pelo método __init__(). É utilizado para inicializar os atributos de um objeto e pode aceitar parâmetros para configurar o objeto quando ele é criado.


```python

class Car:
    def __init__(self, make, model, year):  # Конструктор класса Car
        self.make = make  # Инициализация атрибута make
        self.model = model  # Инициализация атрибута model
        self.year = year  # Инициализация атрибута year

my_car = Car('Toyota', 'Corolla', 2020)  # Создание объекта с использованием конструктора
print(my_car.make)
print(my_car.model)
print(my_car.year)
```

**O que é uma rede de decoradores e como usá-la**? O encadeamento de decoradores em Python é a aplicação sequencial de vários decoradores em uma única função. Cada decorador envolve uma função, adicionando ou alterando seu comportamento. Os decoradores são aplicados de cima para baixo, o que significa que o decorador superior é executado primeiro, depois o próximo e assim por diante. Exemplo de uma rede de decoradores

```python

def decorator_one(func):
    def wrapper(*args, **kwargs):
        print("Decorator One")
        return func(*args, **kwargs)
    return wrapper

def decorator_two(func):
    def wrapper(*args, **kwargs):
        print("Decorator Two")
        return func(*args, **kwargs)
    return wrapper

@decorator_one
@decorator_two
def say_hello():
    print("Hello!")

say_hello()
```

**Set** definido em Python é uma coleção não ordenada de elementos únicos. É usado para armazenar vários elementos sem duplicatas e suporta operações de conjunto como união, interseção e diferença. A criação de um conjunto pode ser feita usando chaves ou a função set():

```python
my_set = {1, 2, 3}
another_set = set([3, 4, 5])
```

Principais características do conjunto:
- Singularidade: todos os elementos de um conjunto são únicos. Adicionar uma duplicata não alterará o conjunto.
- Não ordenado: os elementos não possuem ordem fixa, portanto a indexação não está disponível.
- Mutabilidade: Elementos podem ser adicionados e removidos usando os métodos add() e remove().

Exemplos de operações em conjuntos:

```python
union_set = my_set | another_set  # {1, 2, 3, 4, 5}
intersection_set = my_set & another_set  # {3}
difference_set = my_set - another_set  # {1, 2}

print(union_set)
print(intersection_set)
print(difference_set)

difference_set = another_set - my_set
print(difference_set)
```

**set** é uma ferramenta eficaz para trabalhar com dados exclusivos e realizar operações em conjuntos

Para definir a estrutura da tabela, um estilo declarativo é usado usando **declarative_base**:

```python
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import Column, Integer, String
from sqlalchemy import or_, and_ # Import or_ and and_

Base = declarative_base()

class User(Base):
    __tablename__ = 'users'
    id = Column(Integer, primary_key=True)
    name = Column(String)
```

```python
from sqlalchemy.orm import sessionmaker

Session = sessionmaker(bind=engine)
session = Session()
```

Adicionando e solicitando dados:

```python
# Create the engine and session
engine = create_engine('sqlite:///example.db')

# Create all tables in the database (if they don't exist)
Base.metadata.create_all(engine)  # This line creates the 'users' table

Session = sessionmaker(bind=engine)
session = Session()
```

```python
new_user = User(name='Alice')
session.add(new_user)
session.commit()

users = session.query(User).filter_by(name='Alice').all()
```

No SQLAlchemy, você pode realizar consultas usando diferentes filtros de várias maneiras. Aqui estão alguns exemplos comuns:

**Usando filter()**
O método filter() permite usar expressões mais complexas e combinações de filtros. Você pode usar operadores como ==, !=, >, <, like(), entre outros. Veja um exemplo:

from sqlalchemy import and_, or_  

# Consulta combinando filtros  
users = session.query(User).filter(  
    and_(User.name == 'Alice', User.age > 18)  
).all()  
**Usando filter_by()**
O método filter_by() é um método mais simples que se utiliza de parâmetros nomeados. Ele é ideal para filtros diretos e equivalentes. Exemplo:

# Consulta básica com filtro direto  
users = session.query(User).filter_by(name='Alice', active=True).all()

**Consultas com or_**
Para realizar consultas que atendam a múltiplas condições onde qualquer condição seja verdadeira, você pode usar or_:

from sqlalchemy import or_  

# Consulta usando 'or_'  
users = session.query(User).filter(  
    or_(User.name == 'Alice', User.name == 'Bob')  
).all()

**Filtrando com like()**
Se você quiser filtrar resultados baseados em padrões de string, pode usar o método like():

# Consulta usando like  
users = session.query(User).filter(User.name.like('%Ali%')).all()  # Nomes que contêm "Ali"  

**Exemplo Combinado**
Você pode combinar o uso de filter(), filter_by(), or_, e and_ em uma mesma consulta. Por exemplo:

from sqlalchemy import and_, or_  

# Consulta combinada  
users = session.query(User).filter(  
    or_(  
        User.name.like('%Ali%'),  
        and_(User.age < 20, User.active == True)  
    )  
).all()

Conclusão
Esses são apenas alguns exemplos de como você pode usar filtros diferentes no SQLAlchemy. Dependendo das suas necessidades, você pode combinar filtros conforme necessário para obter os resultados desejados.
